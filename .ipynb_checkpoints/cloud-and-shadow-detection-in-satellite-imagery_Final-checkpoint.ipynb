## Helper Functions

```python
try:
    # Loading the Earth-Sun distance values, using Day of Year (DOY) as the key from given data
    d_values_df = pd.read_csv('/kaggle/input/cloud-and-shadow/d_values.csv')

    # Loading the LISS sensor Exoatmospheric Irradiance (ESUN) values from given data
    liss_esun_df = pd.read_csv('/kaggle/input/cloud-and-shadow/LISS.csv')

    print("Successfully loaded TOA calibration data from d_values.csv and LISS.csv.")

except FileNotFoundError as e:
    print(f"TOA calibration file not found: {e}. Reflectance calculation will not be accurate.")
    d_values_df = None
    liss_esun_df = None


def dn_to_toa_reflectance_accurate(band_array, band_num, julian_day, gain, offset, sun_elevation_angle_deg):
    """
    This function converts Digital Numbers (DN) to accurate Top-of-Atmosphere (TOA) Reflectance
    using the full formula and data from the provided CSV files for the LISS-4 sensor.
    """
    if d_values_df is None or liss_esun_df is None:
        print("Error: Calibration DataFrames are not loaded.")
        return None

    try:
        d = d_values_df.loc[d_values_df['DOY'] == julian_day, 'd'].values[0]
        
        band_str = f'B{band_num}'
        esun = liss_esun_df.loc[liss_esun_df['B'] == band_str, 'LISS4'].values[0]

        # Formula: L = (gain * DN) + offset
        radiance = gain * band_array.astype(np.float64) + offset
        
        # Full Formula: ρ = (π * L * d^2) / (ESUN * sin(θ))
        pi = np.pi
        sun_angle_rad = np.deg2rad(sun_elevation_angle_deg)
        
        numerator = pi * radiance * (d**2)
        denominator = esun * np.sin(sun_angle_rad)
        toa_reflectance = numerator / denominator
        
        return toa_reflectance

    except IndexError:
        print(f"Error: Could not find a value for Julian Day '{julian_day}' or Band 'B{band_num}' in the CSV files.")
        return None
    except Exception as e:
        print(f"An unexpected error occurred during reflectance calculation: {e}")
        return None
```

```python
def parse_metadata_for_scene(scene_path):
    """
    Reads the 'Band_meta.txt' file, parses the key-value pairs, and calculates
    the parameters needed for TOA reflectance conversion, including the satellite ID.
    """
    metadata_filepath = os.path.join(scene_path, 'BAND_META.txt')
    if not os.path.exists(metadata_filepath):
        print(f"Metadata file not found at: {metadata_filepath}")
        return None
        
    meta_dict = {}
    with open(metadata_filepath, 'r') as f:
        for line in f:
            if '=' in line:
                parts = line.split('=', 1)
                key = parts[0].strip()
                value = parts[1].strip()
                meta_dict[key] = value

    try:
        # --- NEW: Extract and map the Satellite ID ---
        sat_id_raw = meta_dict['SatID']
        if 'R2A' in sat_id_raw:
            satellite_category = 'RS2A'
        elif 'R2' in sat_id_raw:
            satellite_category = 'RS2'
        else:
            print(f"Warning: Unknown satellite ID '{sat_id_raw}'. ESUN lookup may fail.")
            satellite_category = None

        # Get Sun Elevation
        sun_elevation = float(meta_dict['SunElevationAtCenter'])

        # Calculate Julian Day (Day of Year) from DateOfPass
        pass_date_str = meta_dict['DateOfPass']
        pass_date_obj = datetime.strptime(pass_date_str, '%d-%b-%Y')
        julian_day = pass_date_obj.timetuple().tm_yday

        # Get Bits Per Pixel for gain calculation
        bits_per_pixel = float(meta_dict['BitsPerPixel'])
        max_dn = (2**bits_per_pixel) - 1

        # --- NEW: Add the satellite category to the top level of the dictionary ---
        final_metadata = {
            'satellite_category': satellite_category
        }
        
        for band_num in [2, 3, 4]:
            lmax = float(meta_dict[f'B{band_num}_Lmax'])
            lmin = float(meta_dict[f'B{band_num}_Lmin'])
            
            gain = (lmax - lmin) / max_dn
            offset = lmin

            final_metadata[f'band{band_num}'] = {
                'julian_day': julian_day,
                'gain': gain,
                'offset': offset,
                'sun_elevation_angle_deg': sun_elevation
            }
        
        return final_metadata

    except (KeyError, ValueError) as e:
        print(f"Error parsing metadata for {scene_path}. Missing or invalid key: {e}")
        return None
```

```python
def check_reflectance_values(scene_path):
    """
    Loads a single band from a scene, gets metadata by parsing
    the Band_meta.txt file, converts it to TOA reflectance, and
    visualizes the value range.
    """
    print("--- Checking TOA Reflectance Value Range using REAL Metadata ---")
    
    scene_metadata = parse_metadata_for_scene(scene_path)
    
    if not scene_metadata:
        print(f"Could not check reflectance for {os.path.basename(scene_path)} due to metadata parsing error.")
        return

    try:
        band_to_check = 4
        band_key = f'band{band_to_check}'
        
        metadata_for_band = scene_metadata[band_key]
        
        gain = metadata_for_band['gain']
        offset = metadata_for_band['offset']
        julian_day = metadata_for_band['julian_day']
        sun_elevation = metadata_for_band['sun_elevation_angle_deg']

        band_path = os.path.join(scene_path, f'BAND{band_to_check}.tif')
        with rasterio.open(band_path) as src:
            raw_dn_band = src.read(1)

        print(f"Converting Band {band_to_check} using parsed metadata...")
        reflectance_band = dn_to_toa_reflectance_accurate(
            band_array=raw_dn_band,
            band_num=band_to_check,
            julian_day=julian_day,
            gain=gain,
            offset=offset,
            sun_elevation_angle_deg=sun_elevation
        )

        if reflectance_band is None:
            print("Reflectance calculation failed.")
            return

        print(f"\nReflectance value stats for BAND {band_to_check}:")
        print(f"  Min value:    {np.min(reflectance_band):.4f}")
        print(f"  Max value:    {np.max(reflectance_band):.4f}")
        print(f"  Mean value:   {np.mean(reflectance_band):.4f}")

        plt.figure(figsize=(10, 6))
        plt.hist(reflectance_band.ravel(), bins=256, color='crimson', range=(0, 1.0))
        plt.title(f'Histogram of TOA Reflectance Values (Band {band_to_check})')
        plt.xlabel('Reflectance')
        plt.ylabel('Pixel Count')
        plt.grid(True, alpha=0.5)
        plt.show()

    except FileNotFoundError:
        print(f"Error: Band file not found at {band_path}")
    except Exception as e:
        print(f"An unexpected error occurred during the check: {e}")
```

```python
def create_pseudo_mask(rgb_chip):
    """
    Generates a pseudo-mask with improved, more robust rules for both
    cloud and shadow detection.
    """
    mask = np.zeros((rgb_chip.shape[0], rgb_chip.shape[1]), dtype=np.uint8)
    
    # --- 1. Cloud Detection ---
    gray = cv2.cvtColor(rgb_chip, cv2.COLOR_RGB2GRAY)
    blue_channel = rgb_chip[:, :, 2]

    cloud_brightness_threshold = 190
    cloud_blue_threshold = 200

    cloud_mask_condition = (gray > cloud_brightness_threshold) & \
                           (blue_channel > cloud_blue_threshold)
    
    mask[cloud_mask_condition] = 1

    # --- 2. Shadow Detection using HSV Color Space ---
    hsv_chip = cv2.cvtColor(rgb_chip, cv2.COLOR_RGB2HSV)
    
    saturation_channel = hsv_chip[:, :, 1]
    
    # --- THIS LINE IS CORRECTED ---
    value_channel = hsv_chip[:, :, 2] # Was incorrectly 'h_sv'

    value_threshold = 70
    saturation_threshold = 80

    shadow_mask_condition = (value_channel < value_threshold) & \
                            (saturation_channel < saturation_threshold) & \
                            (mask != 1)
    
    mask[shadow_mask_condition] = 2
    
    return mask
```

```python
def process_and_chip_scene(scene_path, split, metadata, chip_size=CHIP_SIZE):
    """
    MEMORY-EFFICIENT VERSION: Loads and processes a scene in horizontal strips
    to prevent RAM exhaustion and kernel crashes.
    """
    try:
        b2_src = rasterio.open(os.path.join(scene_path, 'BAND2.tif'))
        b3_src = rasterio.open(os.path.join(scene_path, 'BAND3.tif'))
        b4_src = rasterio.open(os.path.join(scene_path, 'BAND4.tif'))
        
        h, w = b2_src.height, b2_src.width
        scene_name = os.path.basename(os.path.dirname(scene_path))
        chip_count = 0
        
        for i in tqdm(range(0, h, chip_size), desc=f"Chipping {scene_name}", leave=False):
            window = Window(0, i, w, chip_size)

            b2_dn = b2_src.read(1, window=window)
            b3_dn = b3_src.read(1, window=window)
            b4_dn = b4_src.read(1, window=window)

            refl_b2 = dn_to_toa_reflectance_accurate(b2_dn, 2, **metadata['band2'])
            refl_b3 = dn_to_toa_reflectance_accurate(b3_dn, 3, **metadata['band3'])
            refl_b4 = dn_to_toa_reflectance_accurate(b4_dn, 4, **metadata['band4'])

            if refl_b2 is None or refl_b3 is None or refl_b4 is None:
                print(f"Skipping strip in {scene_name} due to reflectance calculation error.")
                continue

            refl_strip = np.stack([refl_b4, refl_b3, refl_b2], axis=-1)
            
            v_min, v_max = np.percentile(refl_strip, (2, 98))
            strip_normalized = np.clip((refl_strip - v_min) / (v_max - v_min + 1e-8), 0, 1)
            strip_8bit = (strip_normalized * 255).astype(np.uint8)

            for j in range(0, w, chip_size):
                if j + chip_size > strip_8bit.shape[1]: continue
                
                chip_img = strip_8bit[:, j:j+chip_size]
                
                if chip_img.shape[0] < chip_size or chip_img.shape[1] < chip_size: continue
                if np.mean(chip_img) < 10: continue

                chip_mask = create_pseudo_mask(chip_img)
                if np.sum(chip_mask == 0) > (chip_size * chip_size * 0.95): continue

                chip_name = f"{scene_name}_{i}_{j}.png"
                img_path = os.path.join(PROCESSED_DATA_DIR, split, "images", chip_name)
                mask_path = os.path.join(PROCESSED_DATA_DIR, split, "masks", chip_name)
                
                cv2.imwrite(img_path, cv2.cvtColor(chip_img, cv2.COLOR_RGB2BGR))
                cv2.imwrite(mask_path, chip_mask)
                chip_count += 1

        b2_src.close()
        b3_src.close()
        b4_src.close()
        gc.collect()
        
        return chip_count

    except Exception as e:
        print(f"Error processing scene {scene_path}: {e}")
        if 'b2_src' in locals() and not b2_src.closed: b2_src.close()
        if 'b3_src' in locals() and not b3_src.closed: b3_src.close()
        if 'b4_src' in locals() and not b4_src.closed: b4_src.close()
        return 0
```

```python
if not scene_paths:
    print("No scene paths found. Skipping data preparation.")
else:
    split_index = int(len(scene_paths) * 0.8)
    train_scenes = scene_paths[:split_index]
    val_scenes = scene_paths[split_index:]

    print("Processing training scenes with REAL TOA Reflectance calculation...")
    total_train_chips = 0
    # The simple sum() comprehension is replaced with a for loop to handle metadata parsing for each scene
    for p in tqdm(train_scenes, desc="Processing Train Scenes"):
        # Get the unique, real metadata for the current scene by calling the parser
        current_metadata = parse_metadata_for_scene(p)
        
        # Proceed only if metadata was successfully parsed
        if current_metadata:
            # The function call now passes the real, unique metadata for the scene
            total_train_chips += process_and_chip_scene(p, 'train', metadata=current_metadata)
        else:
            print(f"Skipping scene due to metadata error: {os.path.basename(p)}")


    print("\nProcessing validation scenes with REAL TOA Reflectance calculation...")
    total_val_chips = 0
    for p in tqdm(val_scenes, desc="Processing Val Scenes"):
        # Get the unique, real metadata for the current scene
        current_metadata = parse_metadata_for_scene(p)

        # Proceed only if metadata was successfully parsed
        if current_metadata:
            # The function call now passes the real, unique metadata for the scene
            total_val_chips += process_and_chip_scene(p, 'val', metadata=current_metadata)
        else:
            print(f"Skipping scene due to metadata error: {os.path.basename(p)}")

    print(f"\nData preparation complete. Generated {total_train_chips} training chips and {total_val_chips} validation chips.")
```

```python
def inspect_training_data(data_dir=PROCESSED_DATA_DIR, num_samples=20):
    """
    Randomly selects and displays image-mask pairs from the training set
    to visually verify the quality of the pseudo-labels.
    """
    print(f"--- Performing Visual Inspection of {num_samples} Random Samples ---")
    
    train_img_dir = os.path.join(data_dir, 'train', 'images')
    train_mask_dir = os.path.join(data_dir, 'train', 'masks')

    if not os.path.exists(train_img_dir) or not os.listdir(train_img_dir):
        print("Training data not found. Cannot perform inspection.")
        return

    # Get a random sample of image files
    sample_files = random.sample(os.listdir(train_img_dir), min(num_samples, len(os.listdir(train_img_dir))))

    for i, file_name in enumerate(sample_files):
        img_path = os.path.join(train_img_dir, file_name)
        mask_path = os.path.join(train_mask_dir, file_name)

        # Read image and mask
        image = cv2.imread(img_path)
        image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
        mask = cv2.imread(mask_path, cv2.IMREAD_GRAYSCALE)

        # Create a plot
        fig, ax = plt.subplots(1, 2, figsize=(12, 6))
        fig.suptitle(f'Sample {i+1}: {file_name}', fontsize=14)
        
        ax[0].imshow(image)
        ax[0].set_title('Image Chip')
        ax[0].axis('off')

        # Display the mask with a color map for clarity
        # We'll use a specific colormap: 0=Black (BG), 1=White (Cloud), 2=Blue (Shadow)
        from matplotlib.colors import ListedColormap
        cmap = ListedColormap(['black', 'white', 'blue'])
        
        cax = ax[1].imshow(mask, cmap=cmap, vmin=0, vmax=2)
        ax[1].set_title('Pseudo-Mask')
        ax[1].axis('off')
        
        # Add a color bar to explain the mask values
        cbar = fig.colorbar(cax, ax=ax[1], ticks=[0, 1, 2], orientation='vertical')
        cbar.ax.set_yticklabels(['0: NoCloud', '1: Cloud', '2: Shadow'])

        plt.show()
```

## PYTORCH DATASET AND DATALOADERS

```python
class CloudDataset(Dataset):
    def __init__(self, data_dir, split='train', transform=None):
        self.img_dir = os.path.join(data_dir, split, 'images')
        self.mask_dir = os.path.join(data_dir, split, 'masks')
        self.image_files = os.listdir(self.img_dir)
        self.transform = transform

    def __len__(self):
        return len(self.image_files)

    def __getitem__(self, idx):
        img_name = self.image_files[idx]
        img_path = os.path.join(self.img_dir, img_name)
        mask_path = os.path.join(self.mask_dir, img_name)
        
        image = cv2.imread(img_path)
        image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
        mask = cv2.imread(mask_path, cv2.IMREAD_GRAYSCALE)
        
        if self.transform:
            augmented = self.transform(image=image, mask=mask)
            image = augmented['image']
            mask = augmented['mask']
            
        return image, mask.long()
```

### Augmentations

```python
train_transform = A.Compose([
    A.HorizontalFlip(p=0.5), A.VerticalFlip(p=0.5), A.RandomRotate90(p=0.5),
    A.Normalize(mean=(0.485, 0.456, 0.406), std=(0.229, 0.224, 0.225)), ToTensorV2(),
])
val_transform = A.Compose([
    A.Normalize(mean=(0.485, 0.456, 0.406), std=(0.229, 0.224, 0.225)), ToTensorV2(),
])
```

### Create Datasets and dataloaders

```python
try:
    train_dataset = CloudDataset(PROCESSED_DATA_DIR, split='train', transform=train_transform)
    val_dataset = CloudDataset(PROCESSED_DATA_DIR, split='val', transform=val_transform)
    train_loader = DataLoader(train_dataset, batch_size=BATCH_SIZE, shuffle=True, num_workers=2, pin_memory=True)
    val_loader = DataLoader(val_dataset, batch_size=BATCH_SIZE, shuffle=False, num_workers=2, pin_memory=True)
    print(f"DataLoaders created. Train batches: {len(train_loader)}, Val batches: {len(val_loader)}")
    run_training = True
except (FileNotFoundError, IndexError):
    print("Processed data not found. Skipping model training.")
    run_training = False
```

## U-Net Model Training

```python
if run_training:
    backbones_to_test = ["resnet34", "resnet50", "efficientnet-b4"]
    
    all_results = []
    all_histories = {}

    for backbone in backbones_to_test:
        print(f"\n{'='*60}\nTraining Model: U-Net with {backbone} backbone\n{'='*60}")
        
        model = smp.Unet(backbone, encoder_weights="imagenet", in_channels=3, classes=NUM_CLASSES).to(DEVICE)
        
        loss_fn = smp.losses.DiceLoss(mode='multiclass', from_logits=True)
        optimizer = torch.optim.AdamW(model.parameters(), lr=LEARNING_RATE)
        lr_scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(optimizer, mode='max', factor=0.1, patience=3)
        scaler = torch.cuda.amp.GradScaler(enabled=(DEVICE == "cuda"))

        history = {k: [] for k in ['train_loss', 'val_iou', 'val_f1', 'val_precision', 'val_recall', 'val_accuracy']}
        best_val_iou = 0.0
        start_time = time.time()

        print(f"Starting U-Net ({backbone}) training for {NUM_EPOCHS} epochs...")
        for epoch in range(NUM_EPOCHS):
            model.train()
            epoch_loss = 0
            progress_bar = tqdm(train_loader, desc=f"Epoch {epoch+1}/{NUM_EPOCHS} [Train]")
            for images, masks in progress_bar:
                images, masks = images.to(DEVICE), masks.to(DEVICE)
                with torch.cuda.amp.autocast(enabled=(DEVICE=="cuda")):
                    outputs = model(images)
                    loss = loss_fn(outputs, masks)
                
                optimizer.zero_grad()
                scaler.scale(loss).backward()
                scaler.step(optimizer)
                scaler.update()
                epoch_loss += loss.item()
                progress_bar.set_postfix(loss=f"{loss.item():.4f}")
            
            # Validation
            model.eval()
            metrics = {k: [] for k in ['iou', 'f1', 'precision', 'recall', 'accuracy']}
            val_progress = tqdm(val_loader, desc=f"Epoch {epoch+1}/{NUM_EPOCHS} [Val]")
            with torch.no_grad():
                for images, masks in val_progress:
                    images, masks = images.to(DEVICE), masks.to(DEVICE)
                    outputs = model(images)
                    pred_masks = outputs.argmax(dim=1)
                    tp, fp, fn, tn = smp.metrics.get_stats(pred_masks, masks, mode='multiclass', num_classes=NUM_CLASSES)
                    
                    metrics['iou'].append(smp.metrics.iou_score(tp, fp, fn, tn, reduction="micro-imagewise").item())
                    metrics['f1'].append(smp.metrics.f1_score(tp, fp, fn, tn, reduction="micro-imagewise").item())
                    metrics['precision'].append(smp.metrics.precision(tp, fp, fn, tn, reduction="micro-imagewise").item())
                    metrics['recall'].append(smp.metrics.recall(tp, fp, fn, tn, reduction="micro-imagewise").item())
                    metrics['accuracy'].append(smp.metrics.accuracy(tp, fp, fn, tn, reduction="micro-imagewise").item())

            history['train_loss'].append(epoch_loss / len(train_loader))
            avg_val_iou = np.mean(metrics['iou'])
            history['val_iou'].append(avg_val_iou)
            history['val_f1'].append(np.mean(metrics['f1']))
            history['val_precision'].append(np.mean(metrics['precision']))
            history['val_recall'].append(np.mean(metrics['recall']))
            history['val_accuracy'].append(np.mean(metrics['accuracy']))
            
            print(f"\nEpoch {epoch+1} Summary: Val IoU: {avg_val_iou:.4f}, Val F1: {np.mean(metrics['f1']):.4f}")
            
            if avg_val_iou > best_val_iou:
                best_val_iou = avg_val_iou
                torch.save(model.state_dict(), f"unet_{backbone}_best.pth")
                print(f"New best model saved with IoU: {best_val_iou:.4f}")
            
            lr_scheduler.step(avg_val_iou)

        print(f"U-Net ({backbone}) training complete.")
        end_time = time.time()
        
        all_histories[f"U-Net ({backbone})"] = history
        all_results.append({
            'Model': f"U-Net ({backbone})",
            'Best Val IoU': best_val_iou,
            'Best Val F1-Score': max(history['val_f1']),
            'Final Val Accuracy': history['val_accuracy'][-1],
            'Training Time (s)': end_time - start_time
        })

    results_df = pd.DataFrame(all_results)
    print("\n\n--- All Model Comparison Results ---")
    display(results_df)

    results_df.to_csv("model_comparison_results.csv", index=False)
    print("\nResults saved to model_comparison_results.csv")

else:
    print("Skipping model training as 'run_training' is False or data is not available.")
```

## PLOTTING TRAINING GRAPHS

```python
def plot_model_comparison(histories_dict):
    """
    Plots the validation IoU of all trained models on a single graph for easy comparison.
    
    """
    if not histories_dict:
        print("No model histories found to plot.")
        return
        
    print("--- Generating Model Comparison Plot ---")
    plt.style.use('seaborn-v0_8-whitegrid')
    plt.figure(figsize=(14, 8))
    
    # Define a list of colors to cycle through for the plots
    colors = plt.cm.viridis(np.linspace(0, 1, len(histories_dict)))

    # Loop through each model's history and plot its validation IoU
    for i, (model_name, history) in enumerate(histories_dict.items()):
        # Ensure there is data to plot
        if 'val_iou' in history and history['val_iou']:
            epochs = range(1, len(history['val_iou']) + 1)
            best_iou = max(history['val_iou'])
            
            # Create a label that includes the model name and its best score
            label = f"{model_name} (Best IoU: {best_iou:.4f})"
            
            plt.plot(epochs, history['val_iou'], 'o-', color=colors[i], label=label)
        
    plt.title('Model Comparison: Validation IoU vs. Epochs', fontsize=16, fontweight='bold')
    plt.xlabel('Epochs', fontsize=12)
    plt.ylabel('Validation IoU Score', fontsize=12)
    plt.legend(fontsize=11)
    plt.xticks(np.arange(1, NUM_EPOCHS + 1, 1)) # Ensure integer ticks for epochs
    plt.tight_layout()
    plt.show()

if 'all_histories' in locals() and all_histories:
    plot_model_comparison(all_histories)
else:
    print("\nCould not generate comparison plot because 'all_histories' was not found or is empty.")
```

## FINAL GIS OUTPUT GENERATION

```python
def generate_gis_outputs(scene_path, unet_model, output_dir=GIS_OUTPUT_DIR):
    print(f"\nStarting GIS output generation for scene: {os.path.basename(os.path.dirname(scene_path))}")
    os.makedirs(output_dir, exist_ok=True)
    
    with rasterio.open(os.path.join(scene_path, 'BAND4.tif')) as src:
        original_crs, original_transform, h, w = src.crs, src.transform, src.height, src.width
        
    full_prediction_mask = np.zeros((h, w), dtype=np.uint8)
    
    b2 = rasterio.open(os.path.join(scene_path, 'BAND2.tif')).read(1)
    b3 = rasterio.open(os.path.join(scene_path, 'BAND3.tif')).read(1)
    b4 = rasterio.open(os.path.join(scene_path, 'BAND4.tif')).read(1)
    rgb_full = np.stack([b4, b3, b2], axis=-1)
    
    v_min, v_max = np.percentile(rgb_full, (2, 98))
    rgb_8bit_full = (np.clip((rgb_full - v_min) / (v_max - v_min + 1e-8), 0, 1) * 255).astype(np.uint8)

    inference_transform = A.Compose([
        A.Normalize(mean=(0.485, 0.456, 0.406), std=(0.229, 0.224, 0.225)), ToTensorV2(),
    ])
    
    unet_model.eval()
    
    for i in tqdm(range(0, h, CHIP_SIZE), desc="Processing Scene Tiles"):
        for j in range(0, w, CHIP_SIZE):
            chip = rgb_8bit_full[i:i+CHIP_SIZE, j:j+CHIP_SIZE]
            if chip.shape[0] < CHIP_SIZE or chip.shape[1] < CHIP_SIZE: continue

            transformed_chip = inference_transform(image=chip)['image']
            with torch.no_grad():
                pred_logits = unet_model(transformed_chip.unsqueeze(0).to(DEVICE))
                pred_chip_mask = torch.argmax(pred_logits, dim=1).squeeze(0).cpu().numpy().astype(np.uint8)
            
            full_prediction_mask[i:i+CHIP_SIZE, j:j+CHIP_SIZE] = pred_chip_mask

    scene_name = os.path.basename(os.path.dirname(scene_path))
    geotiff_path = os.path.join(output_dir, f"{scene_name}_mask.tif")
    print(f"\nSaving 8-bit GeoTIFF mask to: {geotiff_path}")
    with rasterio.open(geotiff_path, 'w', driver='GTiff', height=h, width=w, count=1, dtype=rasterio.uint8, crs=original_crs, transform=original_transform) as dst:
        dst.write(full_prediction_mask, 1)

    results = []
    for class_id, class_name in [(1, "CLOUD"), (2, "SHADOW")]:
        binary_mask = (full_prediction_mask == class_id).astype(np.uint8)
        for geom, val in shapes(binary_mask, transform=original_transform):
            if val == 1:
                results.append({'geometry': shape(geom), 'properties': {'class_name': class_name}})
            
    if results:
        gdf = geopandas.GeoDataFrame.from_features(results)
        gdf.set_crs(crs=original_crs, inplace=True)
        shapefile_path = os.path.join(output_dir, f"{scene_name}_polygons.shp")
        print(f"Saving ESRI Shapefile to: {shapefile_path}")
        gdf.to_file(shapefile_path, driver='ESRI Shapefile')
    
    print("--- GIS Output Generation Complete ---")
```

## Inference Script

```python
if os.path.exists("unet_resnet50_best.pth") and 'val_scenes' in locals() and val_scenes:
    print("\nLoading best U-Net (ResNet50) model for final GIS inference...")
    
    # Creating a model with the SAME architecture as the saved weights.
    unet_model_for_gis = smp.Unet("resnet50", in_channels=3, classes=NUM_CLASSES).to(DEVICE)
    
    # Loading the corresponding weights. This will work without error.
    unet_model_for_gis.load_state_dict(torch.load("unet_resnet50_best.pth"))
    
    # Processing the first available validation scene as a test case
    scene_to_process = val_scenes[0] 
    generate_gis_outputs(scene_to_process, unet_model_for_gis)
else:
    print("\nSkipping final GIS output generation. Trained model 'unet_resnet50_best.pth' or scene data not available.")
```

## Testing

```python
def evaluate_on_test_set(test_set_path, best_model, best_model_name):
    """
    Runs the best trained model on a new test set, performing full TOA reflectance
    conversion using each scene's metadata, evaluates metrics against ground truth masks,
    and generates a final CSV report.
    """
    print(f"--- Starting Evaluation on New Test Set with Model: {best_model_name} ---")

    test_scene_paths = []
    if os.path.exists(test_set_path):
        for scene_id in os.listdir(test_set_path):
            outer_path = os.path.join(test_set_path, scene_id)
            if os.path.isdir(outer_path):
                inner_path = os.path.join(outer_path, scene_id)
                if os.path.isdir(inner_path):
                    test_scene_paths.append(inner_path)
    
    if not test_scene_paths:
        print(f"No valid test scenes found at path: {test_set_path}")
        return None

    results_list = []
    
    best_model.eval()
    inference_transform = A.Compose([
        A.Normalize(mean=(0.485, 0.456, 0.406), std=(0.229, 0.224, 0.225)),
        ToTensorV2(),
    ])

    for scene_path in tqdm(test_scene_paths, desc="Evaluating Test Scenes"):
        registration_id = os.path.basename(os.path.dirname(scene_path))
        
        current_metadata = parse_metadata_for_scene(scene_path)
        if not current_metadata:
            print(f"Warning: Metadata not found for {registration_id}, skipping.")
            continue
            
        mask_path = os.path.join(scene_path, "mask.tif") 
        if not os.path.exists(mask_path):
            print(f"Warning: Ground truth mask not found for {registration_id}, skipping.")
            continue
        with rasterio.open(mask_path) as src:
            ground_truth_mask = src.read(1)
            
        h, w = ground_truth_mask.shape

        predicted_mask = np.zeros((h, w), dtype=np.uint8)
        
        try:
            b2_dn = rasterio.open(os.path.join(scene_path, 'BAND2.tif')).read(1)
            b3_dn = rasterio.open(os.path.join(scene_path, 'BAND3.tif')).read(1)
            b4_dn = rasterio.open(os.path.join(scene_path, 'BAND4.tif')).read(1)

            refl_b2 = dn_to_toa_reflectance_accurate(b2_dn, 2, **current_metadata['band2'])
            refl_b3 = dn_to_toa_reflectance_accurate(b3_dn, 3, **current_metadata['band3'])
            refl_b4 = dn_to_toa_reflectance_accurate(b4_dn, 4, **current_metadata['band4'])
            
            refl_full = np.stack([refl_b4, refl_b3, refl_b2], axis=-1)
            
            v_min, v_max = np.percentile(refl_full, (2, 98))
            rgb_8bit_full = (np.clip((refl_full - v_min) / (v_max - v_min + 1e-8), 0, 1) * 255).astype(np.uint8)

            for i in range(0, h, CHIP_SIZE):
                for j in range(0, w, CHIP_SIZE):
                    chip = rgb_8bit_full[i:i+CHIP_SIZE, j:j+CHIP_SIZE]
                    if chip.shape[0] < CHIP_SIZE or chip.shape[1] < CHIP_SIZE: continue

                    transformed_chip = inference_transform(image=chip)['image']
                    with torch.no_grad():
                        pred_logits = best_model(transformed_chip.unsqueeze(0).to(DEVICE))
                        pred_chip_mask = torch.argmax(pred_logits, dim=1).squeeze(0).cpu().numpy().astype(np.uint8)
                    predicted_mask[i:i+CHIP_SIZE, j:j+CHIP_SIZE] = pred_chip_mask
        except Exception as e:
            print(f"Error during prediction for {registration_id}: {e}")
            continue

        tp, fp, fn, tn = smp.metrics.get_stats(torch.from_numpy(predicted_mask), torch.from_numpy(ground_truth_mask), mode='multiclass', num_classes=NUM_CLASSES)
        
        results_list.append({
            'Registration Id': registration_id,
            'IOU': smp.metrics.iou_score(tp, fp, fn, tn, reduction="micro").item(),
            'F1-Score': smp.metrics.f1_score(tp, fp, fn, tn, reduction="micro").item(),
            'Precision': smp.metrics.precision(tp, fp, fn, tn, reduction="micro").item(),
            'Recall': smp.metrics.recall(tp, fp, fn, tn, reduction="micro").item(),
            'Accuracy': smp.metrics.accuracy(tp, fp, fn, tn, reduction="micro").item()
        })

    if not results_list:
        print("No results were generated. The test run might have failed for all scenes.")
        return None
        
    results_df = pd.DataFrame(results_list)
    output_csv_path = "/kaggle/working/Final_Test_Set_Results.csv"
    results_df.to_csv(output_csv_path, index=False)
    
    print(f"\nTest set evaluation complete. Results saved to: {output_csv_path}")
    print("Final Test Metrics (Averaged Across All Test Scenes):")
    print(results_df.mean(numeric_only=True))
    
    return results_df
```

```python
final_test_set_folder = "/kaggle/input/cloud-and-shadows-test"

if 'results_df' in locals() and not results_df.empty:
    best_model_row = results_df.loc[results_df['Best Val IoU'].idxmax()]
    best_model_name = best_model_row['Model']
    best_model_backbone = best_model_name.split('(')[1].split(')')[0]
    best_model_weights_path = f"unet_{best_model_backbone}_best.pth"

    print(f"\n--- Found best performing model: {best_model_name} ---")

    if os.path.exists(best_model_weights_path):
        print(f"Loading weights from: {best_model_weights_path}")
        
        best_model = smp.Unet(best_model_backbone, in_channels=3, classes=NUM_CLASSES).to(DEVICE)
        best_model.load_state_dict(torch.load(best_model_weights_path))
        
        
        # Generate GIS Outputs
        print("\n--- Running Inference to Generate GIS Outputs ---")
        test_scene_paths = []
        if os.path.exists(final_test_set_folder):
            for scene_id in os.listdir(final_test_set_folder):
                outer_path = os.path.join(final_test_set_folder, scene_id)
                if os.path.isdir(outer_path):
                    inner_path = os.path.join(outer_path, scene_id)
                    if os.path.isdir(inner_path):
                        test_scene_paths.append(inner_path)
        
        if not test_scene_paths:
            print(f"No valid test scenes found at path: {final_test_set_folder}")
        else:
            print(f"\nFound {len(test_scene_paths)} scenes to process in the test set.")
            
            for scene_path in tqdm(test_scene_paths, desc="Processing Test Scenes for GIS"):
                generate_gis_outputs(scene_path, best_model)
            
            print("\nInference on all test scenes is complete.")
            print(f"Your final GeoTIFF and Shapefile outputs are saved in the '{GIS_OUTPUT_DIR}' directory.")
            
    else:
        print(f"Error: Weight file for the best model not found at '{best_model_weights_path}'")
else:
    print("\nCould not run final inference. Model comparison results DataFrame ('results_df') not found.")
```
